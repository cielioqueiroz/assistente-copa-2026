# Copa 2026 AI - Demonstracao

Demonstracao do assistente virtual sobre a Copa do Mundo 2026. Este notebook usa o mesmo codigo e a mesma base de conhecimento do projeto (`src/assistente.py` e `data/base_conhecimento.json`).

Se a variavel de ambiente `GEMINI_API_KEY` estiver configurada, o assistente responde usando o modelo Gemini (IA generativa). Caso contrario, ele usa a busca local por palavras-chave. As perguntas e o codigo sao os mesmos nos dois modos.

## 1. Carregar o assistente

```python
import sys
import os

sys.path.append(os.path.join(os.getcwd(), "src"))

from assistente import AssistenteCopa2026

assistente = AssistenteCopa2026()
modo = "IA generativa" if assistente.usando_ia() else "busca local"
print(f"{assistente.nome} v{assistente.versao} carregado.")
print(f"Modo ativo: {modo}.")
```

## 2. Perguntas de exemplo

As quatro primeiras vem da base de conhecimento. A ultima (tecnico atual) nao esta na base: no modo de IA o assistente busca a resposta na web.

```python
perguntas = [
    "Qual o grupo do Brasil e quem sao os adversarios?",
    "E o grupo da Argentina?",
    "Quem sao os favoritos?",
    "Quando e onde sera a final?",
    "Quem e o atual tecnico da selecao brasileira?",
]

for pergunta in perguntas:
    print(f"Voce: {pergunta}")
    print("Assistente:")
    print(assistente.responder(pergunta))
    print("-" * 50)
```

## 3. Estatisticas da conversa

```python
total = len(assistente.historico)
interacoes = assistente.contar_interacoes()

print(f"Total de mensagens no historico: {total}")
print(f"Perguntas feitas pelo usuario: {interacoes}")
```

## 4. Historico completo

```python
for i, msg in enumerate(assistente.historico, 1):
    tipo = "Usuario" if msg["tipo"] == "usuario" else "Assistente"
    hora = msg["timestamp"].split("T")[1][:8]
    conteudo = msg["conteudo"]
    if len(conteudo) > 100:
        conteudo = conteudo[:100] + "..."
    print(f"[{i}] {tipo} ({hora})")
    print(conteudo)
    print("-" * 50)
```

## Conclusao

O notebook mostrou o assistente respondendo perguntas sobre selecoes, favoritos, jogadores e datas, alem de avisar quando nao tem o dado pedido. Todas as respostas vem da base de conhecimento em JSON.
